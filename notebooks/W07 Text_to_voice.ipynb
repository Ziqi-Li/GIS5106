### Voice to Text
This is a mini example of showing how to use an open-source sound model to generate audio files from text.

Link: https://huggingface.co/hexgrad/Kokoro-82M

```python
from kokoro import KPipeline
from IPython.display import display, Audio
import soundfile as sf
import torch
pipeline = KPipeline(lang_code='a')
```

#### Your text

```python
text = '''
Geography is more than maps and places.
It studies the relationship between people and their environments,
revealing the spatial dimensions of global change, urban development,
climate risks, and social inequality.
To study geography is to understand how the world works—and to use spatial thinking to solve real-world problems.
'''
```

```python
generator = pipeline(text, voice='af_heart')
for i, (gs, ps, audio) in enumerate(generator):
    print(i, gs, ps)
    display(Audio(data=audio, rate=24000, autoplay=i==0))
    sf.write(f'{i}.wav', audio, 24000)
```
